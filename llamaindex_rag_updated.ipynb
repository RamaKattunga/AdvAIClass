## ✅ What is RAG (Retrieval Augmented Generation)?
- Augment LLMs with private data (e.g., files, APIs, databases).
- Solve limitations of LLM knowledge cutoff.
- LlamaIndex helps with ingestion, indexing, and querying.

## ✅ Components of LlamaIndex
1. **Data Loading** – Load PDFs, Slack logs, etc.
2. **Indexing** – Parse and structure the data.
3. **Query Interface** – Retrieve relevant context and synthesize responses.

```python
# ✅ Sample RAG Setup Using LlamaIndex
!pip install llama-index --upgrade

from llama_index.core import VectorStoreIndex, SimpleDirectoryReader

# Load documents
documents = SimpleDirectoryReader('data').load_data()

# Create index
index = VectorStoreIndex.from_documents(documents)

# Query
query_engine = index.as_query_engine()
response = query_engine.query("What is LlamaIndex?")
print(response)
```

## 📌 Techniques to Improve RAG Performance
### 1. Metadata Augmentation
- Add context like page number, section title, etc.

### 2. Decoupling Embeddings
- Embed summaries or filtered content instead of raw text.

### 3. Structured Retrieval
- Organize documents with tags (e.g., year, department).
- Filter during retrieval using metadata.

### 4. Sentence-Level Embedding
- Embed sentences individually and expand surrounding context during synthesis.

## 🔧 Fine-Tuning
- Fine-tune **embedding models** using synthetic Q&A pairs.
- Fine-tune **LLMs** (e.g., distill GPT-4 CoT to GPT-3.5).

LlamaIndex provides tools to streamline these processes.
