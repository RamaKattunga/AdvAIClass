## Chapter 1: Tokenization

**What is Tokenization?**
Tokenization is the process of breaking down text into smaller units called **tokens**, such as words, sentences, or phrases. It's the first step in NLP, preparing text for further analysis by converting a continuous string into manageable pieces.

**Why is it important?**
- Tokens are the building blocks for tasks like text classification or sentiment analysis.
- It helps standardize text by splitting it into consistent units.

**Example**: The sentence "I love coding!" can be tokenized into words: ["I", "love", "coding", "!"]

In this section, we'll use NLTK's `word_tokenize` and `sent_tokenize` to perform word and sentence tokenization.

**Installation** (run this once if NLTK is not installed):


````python
# Import NLTK and download required data
import nltk
nltk.download('punkt')  # Download the tokenizer models

# Sample text
text = "I love coding! Python is fun. Let's learn NLP."

# Word Tokenization
word_tokens = nltk.word_tokenize(text)
print("Word Tokens:", word_tokens)

# Sentence Tokenization
sentence_tokens = nltk.sent_tokenize(text)
print("Sentence Tokens:", sentence_tokens)
````

**Code Explanation**:
- `nltk.download('punkt')`: Downloads NLTK's pre-trained tokenization models.
- `word_tokenize`: Splits text into individual words and punctuation.
- `sent_tokenize`: Splits text into sentences based on punctuation and grammar.

## Chapter 2: Part-of-Speech (POS) Tagging

**What is POS Tagging?**
POS tagging assigns a part-of-speech tag (e.g., noun, verb, adjective) to each token in a sentence. This helps understand the grammatical structure of text, which is useful for tasks like text analysis or machine translation.

**Why is it important?**
- Identifies the role of each word in a sentence (e.g., is "run" a verb or a noun?).
- Enables more advanced NLP tasks like dependency parsing or NER.

**Example**: For "I love coding", POS tags might be: [("I", "PRP"), ("love", "VBP"), ("coding", "NN")], where PRP = pronoun, VBP = verb, NN = noun.

We'll use NLTK's `pos_tag` function, which requires the `averaged_perceptron_tagger` model.

````python
# Download POS tagger model
nltk.download('averaged_perceptron_tagger')

# Sample text
text = "I love coding in Python"
tokens = nltk.word_tokenize(text)

# POS Tagging
pos_tags = nltk.pos_tag(tokens)
print("POS Tags:", pos_tags)

# Common POS tags reference
print("\nCommon POS Tags:")
print("NN: Noun, VB: Verb, JJ: Adjective, PRP: Pronoun")
````

**Code Explanation**:
- `pos_tag`: Assigns a POS tag to each token based on its context in the sentence.
- Tags follow the Penn Treebank tagset (e.g., NN for noun, VB for verb).

## Chapter 3: Named Entity Recognition (NER)

**What is NER?**
Named Entity Recognition identifies and classifies named entities (e.g., people, organizations, locations) in text. It's used in applications like information extraction and question answering.

**Why is it important?**
- Extracts key information from text (e.g., identifying "Apple" as an organization).
- Supports tasks like building knowledge graphs or chatbots.

**Example**: For "Elon Musk founded Tesla", NER might identify: Elon Musk (PERSON), Tesla (ORGANIZATION).

We'll use NLTK's `ne_chunk` with the `maxent_ne_chunker` and `words` models.

````python
# Download NER models
nltk.download('maxent_ne_chunker')
nltk.download('words')

# Sample text
text = "Elon Musk founded Tesla in California"
tokens = nltk.word_tokenize(text)
pos_tags = nltk.pos_tag(tokens)

# NER
ner_tree = nltk.ne_chunk(pos_tags)
print("NER Output:")
for subtree in ner_tree:
    if isinstance(subtree, nltk.Tree):
        print(f"Entity: {' '.join(word for word, tag in subtree.leaves())}, Type: {subtree.label()}")
````

**Code Explanation**:
- `ne_chunk`: Labels tokens as named entities (e.g., PERSON, ORGANIZATION, GPE for geopolitical entity).
- The output is a tree structure; we extract entities by checking for `nltk.Tree` objects.

## Chapter 4: Stop Words Removal, Stemming, and Lemmatization

**What are Stop Words?**
Stop words are common words (e.g., "the", "is", "and") that carry little meaning and are often removed to focus on content words.

**What is Stemming?**
Stemming reduces words to their root form by removing suffixes (e.g., "running" → "run"). It's fast but can be imprecise.

**What is Lemmatization?**
Lemmatization reduces words to their base form (lemma) based on their part of speech (e.g., "better" → "good"). It's more accurate but slower.

**Why are they important?**
- Stop words removal reduces noise in text data.
- Stemming and lemmatization normalize text, improving consistency for tasks like text classification.

We'll use NLTK's stop words list, `PorterStemmer`, and `WordNetLemmatizer`.

````python
# Download stop words and lemmatizer data
nltk.download('stopwords')
nltk.download('wordnet')

from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer

# Sample text
text = "The quick brown foxes are running faster"
tokens = nltk.word_tokenize(text.lower())  # Convert to lowercase

# Stop Words Removal
stop_words = set(stopwords.words('english'))
filtered_tokens = [word for word in tokens if word not in stop_words]
print("After Stop Words Removal:", filtered_tokens)

# Stemming
stemmer = PorterStemmer()
stemmed_tokens = [stemmer.stem(word) for word in filtered_tokens]
print("After Stemming:", stemmed_tokens)

# Lemmatization
lemmatizer = WordNetLemmatizer()
lemmatized_tokens = [lemmatizer.lemmatize(word) for word in filtered_tokens]
print("After Lemmatization:", lemmatized_tokens)
````

**Code Explanation**:
- `stopwords.words('english')`: Provides a list of English stop words.
- `PorterStemmer`: Applies rule-based stemming (e.g., removes "-ing").
- `WordNetLemmatizer`: Uses WordNet's lexical database for accurate lemmatization.

## Chapter 5: Text Summarization and Vectorization

**What is Text Summarization?**
Text summarization generates a concise summary of a longer text, either extractively (selecting key sentences) or abstractively (generating new sentences). We'll focus on **TextRank**, an extractive method.

**What are Vectors in Summarization?**
Text must be converted into numerical vectors for algorithms to process. Common methods include:
- **Bag of Words (BoW)**: Represents text as a vector of word frequencies.
- **TF-IDF**: Weights words based on their importance in a document relative to a corpus.
- **Word Embeddings**: Dense vectors capturing semantic meaning (we'll use a simple example).

**What is TextRank?**
TextRank is a graph-based algorithm that ranks sentences based on their similarity, similar to PageRank.

We'll implement a simple TextRank summarizer and demonstrate BoW and TF-IDF using scikit-learn.

````python
# Install scikit-learn if needed
# !pip install scikit-learn

from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
import numpy as np

# Sample documents
docs = [
    "I love coding in Python",
    "Python is great for NLP",
    "Coding is fun"
]

# Bag of Words
bow_vectorizer = CountVectorizer()
bow_matrix = bow_vectorizer.fit_transform(docs)
print("BoW Vocabulary:", bow_vectorizer.get_feature_names_out())
print("BoW Matrix:\n", bow_matrix.toarray())

# TF-IDF
tfidf_vectorizer = TfidfVectorizer()
tfidf_matrix = tfidf_vectorizer.fit_transform(docs)
print("\nTF-IDF Vocabulary:", tfidf_vectorizer.get_feature_names_out())
print("TF-IDF Matrix:\n", tfidf_matrix.toarray())

# Simple TextRank Summarizer
def textrank_summary(text, num_sentences=2):
    sentences = nltk.sent_tokenize(text)
    if len(sentences) <= num_sentences:
        return text
    
    # Create TF-IDF matrix for sentences
    tfidf = TfidfVectorizer().fit_transform(sentences)
    # Compute similarity matrix
    similarity_matrix = (tfidf * tfidf.T).toarray()
    # Rank sentences using PageRank
    scores = np.array([sum(row) for row in similarity_matrix])
    top_indices = scores.argsort()[-num_sentences:][::-1]
    summary = ' '.join([sentences[i] for i in sorted(top_indices)])
    return summary

# Sample text for summarization
text = "Natural language processing is a field of AI. It enables computers to understand human language. Python is widely used in NLP. NLTK is a popular library for NLP tasks."
summary = textrank_summary(text)
print("\nTextRank Summary:", summary)
````

**Code Explanation**:
- `CountVectorizer`: Creates a BoW matrix where each element is the frequency of a word.
- `TfidfVectorizer`: Creates a TF-IDF matrix, weighting words by their importance.
- `textrank_summary`: Implements a basic TextRank algorithm by ranking sentences based on TF-IDF similarity.

## Chapter 6: Topic Modeling

**What is Topic Modeling?**
Topic modeling identifies latent topics in a collection of documents. It’s useful for discovering themes in large text corpora (e.g., news articles, reviews).

**How does it work?**
- Algorithms like **Latent Dirichlet Allocation (LDA)** assume documents are mixtures of topics, and topics are mixtures of words.
- We’ll use NLTK for preprocessing and scikit-learn for LDA.

**Example**: In a set of articles, LDA might identify topics like "technology" or "sports" based on word distributions.

````python
from sklearn.decomposition import LatentDirichletAllocation

# Sample documents
docs = [
    "Python is great for coding and NLP",
    "Football is a popular sport",
    "Coding in Python is fun",
    "Sports like football are exciting"
]

# Preprocess and vectorize
vectorizer = CountVectorizer(stop_words='english')
doc_matrix = vectorizer.fit_transform(docs)

# Apply LDA
lda = LatentDirichletAllocation(n_components=2, random_state=42)
lda.fit(doc_matrix)

# Display topics
words = vectorizer.get_feature_names_out()
for i, topic in enumerate(lda.components_):
    top_words = [words[j] for j in topic.argsort()[-5:]]
    print(f"Topic {i+1}: {top_words}")
````

**Code Explanation**:
- `CountVectorizer`: Converts documents to a BoW matrix, removing stop words.
- `LatentDirichletAllocation`: Fits an LDA model to identify topics.
- Each topic is represented by its top words, indicating the theme.

## Chapter 7: Class Imbalance, ROC, AUC, and F1 Score (Part 1)

**What is Class Imbalance?**
Class imbalance occurs when one class (e.g., positive reviews) is much more frequent than another (e.g., negative reviews) in a dataset. This can bias models toward the majority class.

**What are ROC, AUC, and F1 Score?**
- **ROC (Receiver Operating Characteristic)**: A curve plotting True Positive Rate vs. False Positive Rate at various thresholds.
- **AUC (Area Under the Curve)**: Measures the overall performance of a classifier (higher is better).
- **F1 Score**: Harmonic mean of precision and recall, useful for imbalanced datasets.

**Why are they important?**
- They help evaluate text classification models, especially when classes are imbalanced.

We'll build a simple text classifier using scikit-learn and compute these metrics.

````python
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import roc_curve, auc, f1_score, classification_report
import matplotlib.pyplot as plt

# Sample dataset (simulating reviews)
texts = [
    "I love this product", "Great experience", "Amazing quality",  # Positive
    "Terrible service", "Not good"  # Negative
]
labels = [1, 1, 1, 0, 0]  # 1 = Positive, 0 = Negative

# Split data
X_train, X_test, y_train, y_test = train_test_split(texts, labels, test_size=0.4, random_state=42)

# Vectorize text
vectorizer = TfidfVectorizer()
X_train_vec = vectorizer.fit_transform(X_train)
X_test_vec = vectorizer.transform(X_test)

# Train classifier
clf = LogisticRegression()
clf.fit(X_train_vec, y_train)

# Predict probabilities
y_scores = clf.predict_proba(X_test_vec)[:, 1]
y_pred = clf.predict(X_test_vec)

# Compute ROC and AUC
fpr, tpr, _ = roc_curve(y_test, y_scores)
roc_auc = auc(fpr, tpr)

# Plot ROC curve
plt.figure()
plt.plot(fpr, tpr, label=f'ROC curve (AUC = {roc_auc:.2f})')
plt.plot([0, 1], [0, 1], 'k--')
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('ROC Curve')
plt.legend()
plt.show()

# Compute F1 Score
f1 = f1_score(y_test, y_pred)
print(f"F1 Score: {f1:.2f}")
print("\nClassification Report:\n", classification_report(y_test, y_pred))
````

**Code Explanation**:
- `TfidfVectorizer`: Converts text to TF-IDF vectors for classification.
- `LogisticRegression`: A simple classifier for binary classification.
- `roc_curve` and `auc`: Compute the ROC curve and AUC score.
- `f1_score`: Measures the balance between precision and recall.
