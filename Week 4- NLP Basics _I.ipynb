### Importing Required Libraries

These imports are foundational for NLP preprocessing tasks.

````python
# Import necessary libraries
import nltk
import string
import re
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.stem import PorterStemmer, LancasterStemmer, SnowballStemmer, WordNetLemmatizer
from nltk.corpus import stopwords
from nltk.probability import FreqDist
from nltk import pos_tag, ne_chunk
from nltk.chunk import RegexpParser
````

###  Tokenization

Tokenization is the process of breaking a sentence or document into smaller units:

- `word_tokenize`: Splits text into individual words (tokens).
- `sent_tokenize`: Splits text into sentences.

We demonstrate this on a sample sentence to extract both word-level and sentence-level tokens.

````python
# 1. Tokenization
# Tokenization is the process of breaking down text into smaller units, like words or sentences.
text = "Today we will understand tokenization. It is a process of breaking up strings into tokens."
words = word_tokenize(text)
print("Tokenization (words):", words)
sentences = sent_tokenize(text)
print("Tokenization (sentences):", sentences)
````

### Removing Punctuation

Text often includes punctuation which may not be useful for analysis. We use a regex:

- `re.sub(r'[\W]', '', text)`: Removes all non-alphanumeric characters except whitespace.

This helps simplify the text for further NLP tasks like vectorization or keyword extraction.

````python
# 2. Removing Punctuation
# Punctuation is often removed during text preprocessing to focus on meaningful words.
text_with_punct = "Hello, world! This is a test."
text_no_punct = re.sub(r'[^\w\s]', '', text_with_punct)
print("Text without punctuation:", text_no_punct)
````

### Stop Words Removal

Stop words are common words (like "is", "the", "and") that carry little useful meaning.

- We retrieve stop words using `stopwords.words('english')`.
- After tokenizing the sentence, we filter out both stop words and punctuation.
- This reduces noise and improves the quality of our NLP features.

This step is crucial for improving the signal-to-noise ratio in text data.

````python
# 3. Stop Words Removal
# Stop words are common words (like "the", "is") that are often removed as they don't add much meaning.
stop_words = set(stopwords.words('english'))
text = "This is a sample sentence, showing off the stop words filtration."
words = word_tokenize(text)
filtered_words = [word for word in words if word.lower() not in stop_words and word not in string.punctuation]
print("Filtered words (no stop words and punctuation):", filtered_words)
````

```markdown
###  Stemming

Stemming is the process of reducing words to their base or root form.

- This cell shows different stemmers:
  - `PorterStemmer`: Popular and widely used.
  - `LancasterStemmer`: More aggressive than Porter.
  - `SnowballStemmer`: A newer, multilingual stemmer.

The stemmers are applied to sample words to demonstrate how inflectional forms are reduced to stems.
```

````python
# 4. Stemming
# Stemming reduces words to their root form by removing suffixes.
porter = PorterStemmer()
lancaster = LancasterStemmer()
snowball = SnowballStemmer("english")
words_to_stem = ["give", "giving", "given", "gave"]
print("Porter Stemmer:", [porter.stem(word) for word in words_to_stem])
print("Lancaster Stemmer:", [lancaster.stem(word) for word in words_to_stem])
print("Snowball Stemmer:", [snowball.stem(word) for word in words_to_stem])
````

###  Lemmatization

Lemmatization is similar to stemming but returns **actual words** instead of just stems.

- Uses `WordNetLemmatizer` from NLTK.
- Requires the correct part of speech (POS) for better accuracy.
- The word "better" is lemmatized to "good" with POS='a' (adjective).

More accurate than stemming for linguistic analysis.

````python
# 5. Lemmatization
# Lemmatization is similar to stemming but uses a dictionary to ensure the root is a valid word.
lemmatizer = WordNetLemmatizer()
sentence = "The dogs are running quickly."

# Tokenize and POS tag
words = word_tokenize(sentence)
tagged_words = pos_tag(words)

# Function to map NLTK POS tags to WordNet POS tags
def get_wordnet_pos(nltk_tag):
    if nltk_tag.startswith('N'):
        return 'n'  # Noun
    elif nltk_tag.startswith('V'):
        return 'v'  # Verb
    elif nltk_tag.startswith('J'):
        return 'a'  # Adjective
    elif nltk_tag.startswith('R'):
        return 'r'  # Adverb
    else:
        return 'n'  # Default to noun if unknown

# Lemmatize with mapped POS tags
lemmatized_words = [lemmatizer.lemmatize(word, pos=get_wordnet_pos(tag)) for word, tag in tagged_words]
print("Lemmatization:", lemmatized_words)
````

### POS Tagging

POS tagging assigns a part of speech (noun, verb, adjective, etc.) to each word.

- `pos_tag()` returns a list of tuples (word, POS).
- Helps in parsing grammatical structure.
- Useful for lemmatization, chunking, and syntactic analysis.

Enables more intelligent NLP by understanding grammatical roles.

````python
# 6. Parts of Speech (POS) Tagging
# POS tagging assigns a part of speech (like noun, verb) to each word in a sentence.
sentence = "The cat sat on the mat."
words = word_tokenize(sentence)
tagged = pos_tag(words)
print("POS Tagging:", tagged)
````

````python
# 7. Named Entity Recognition (NER)
# NER identifies named entities like persons, organizations, or locations in text.
sentence = "The US President stays in the White House."
words = word_tokenize(sentence)
tagged = pos_tag(words)
ner = ne_chunk(tagged)
print("NER:", ner)
````

````python
# 8. Chunking
# Chunking groups words into larger units, like noun phrases, based on grammatical rules.
sentence = "The big cat ate the little mouse who was after the fresh cheese."
words = word_tokenize(sentence)
tagged = pos_tag(words)
grammar = "NP: {<DT>?<JJ>*<NN>}"  # Noun Phrase: Determiner (DT), Adjective (JJ), Noun (NN)
cp = RegexpParser(grammar)
chunked = cp.parse(tagged)
print("Chunking:", chunked)
````

````python

# 9. Frequency Distribution
# Frequency distribution shows how often each word appears in the text.
text = "This is a sample sentence. This sentence is just for demonstration."
words = word_tokenize(text)
fdist = FreqDist(words)
print("Frequency Distribution (top 5 words):", fdist.most_common(5))
````
